## other structures 

### sparse

```python
import tensorflow as tf

sparse = tf.SparseTensor(indices=[[0, 2], [1, 1], [2, 0]],
                         values=[1, 2, 3],
                         dense_shape=[3, 3])

print("Sparse tensor:", sparse)
print("Convert back to dense:\n", tf.sparse.to_dense(sparse))
```

### stack

```python
ta = tf.TensorArray(dtype=tf.float32, size=3)

# Write values
ta = ta.write(0, [1.0, 2.0])
ta = ta.write(1, [3.0, 4.0])
ta = ta.write(2, [5.0, 6.0])

# Stack back into a single tensor
stacked = ta.stack()
print(stacked)
# Output: [[1, 2], [3, 4], [5, 6]]
```

### ragged (different lenghts)

```python
rt = tf.ragged.constant([[1, 2, 3],
                         [4, 5],
                         [6]])
print(rt)
# Output: <tf.RaggedTensor [[1, 2, 3], [4, 5], [6]]>

print(rt[1])  # Row 1 = [4, 5]
```

### strings

```python
st = tf.constant(["hello", "café"])  
print(st)  # UTF-8 encoded internally

# String ops
upper = tf.strings.upper(st)
print(upper.numpy())  # [b'HELLO' b'CAF\xc3\x89']
```

### join


```python
s1 = tf.constant([[1, 2], [3, 4]])
s2 = tf.constant([[2, 3], [4, 5]])

union = tf.sets.union(s1, s2)
print("Union:", tf.sparse.to_dense(union))
# Output: [[1 2 3], [3 4 5]]
```

## custom loss

```python
def huber_fn(y_true, y_pred):
 error = y_true - y_pred
 is_small_error = tf.abs(error) < 1
 squared_loss = tf.square(error) / 2
 linear_loss  = tf.abs(error) - 0.5
 return tf.where(is_small_error, squared_loss, linear_loss)
```

```python
import tensorflow as tf

y_true = tf.constant([1.0, 2.0, 3.0])   # ground truth
y_pred = tf.constant([1.5, 2.5, 2.0])   # predictions

loss = huber_fn(y_true, y_pred)
print(loss)
```

## Custom Activation Functions, Regularizers,and Constraints

```python
 def my_softplus(z):
     return tf.math.log(1.0 + tf.exp(z))
     
 def my_l1_regularizer(weights):
     return tf.reduce_sum(tf.abs(0.01 * weights))
     
 def my_positive_weights(weights):  # return value is just tf.nn.relu(weights)
     return tf.where(weights < 0., tf.zeros_like(weights), weights)
```
